```python
import nltk
from nltk.corpus import stopwords
from collections import Counter
import re

# Download NLTK stopwords data if not already downloaded
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
```

```python
# Function to preprocess text and extract keywords
def extract_keywords(text):
    if pd.isna(text):  # Handle NaN values
        return []
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers
    words = text.split()
    # Filter out stopwords and single-character words
    keywords = [word for word in words if word not in stop_words and len(word) > 1]
    return keywords

# Apply the function to the 'Title' column and flatten the list of keywords
all_keywords = df_filtered['Title'].apply(extract_keywords).explode()

# Count the frequency of each keyword
keyword_counts = Counter(all_keywords)

# Get the top 20 most common keywords
top_keywords = pd.DataFrame(keyword_counts.most_common(20), columns=['Keyword', 'Count'])
display(top_keywords)
```

## Prepare Authors for Co-authorship Analysis

### Subtask:
Clean and prepare the 'Authors' column by splitting author strings and handling multiple authors per entry.


**Reasoning**:
First, I will define a function to clean author strings as specified, splitting by semicolon, stripping whitespace, and handling empty inputs to ensure a clean list of author names for each entry.



```python
def clean_authors(authors_string):
    if pd.isna(authors_string) or not isinstance(authors_string, str) or authors_string.strip() == '':
        return []
    # Split by semicolon, strip whitespace from each author, and filter out empty strings
    return [author.strip() for author in authors_string.split(';') if author.strip()]

print("Defined the 'clean_authors' function.")
```

**Reasoning**:
Now that the `clean_authors` function is defined, I will apply it to the 'Authors' column of the `df_filtered` DataFrame to create a new 'cleaned_authors' column, which will contain a list of individual authors for each entry.



```python
df_filtered['cleaned_authors'] = df_filtered['Authors'].apply(clean_authors)
display(df_filtered[['Authors', 'cleaned_authors']].head())
```

## Generate Co-authorship Pairs

### Subtask:
For each publication, identify all unique pairs of authors who have collaborated.


**Reasoning**:
I need to import the `combinations` function from the `itertools` module to generate all possible unique pairs of authors from each publication. Then, I will initialize an empty list to store these pairs.



```python
from itertools import combinations

co_authorship_pairs = []

print("Imported 'combinations' and initialized 'co_authorship_pairs' list.")
```

**Reasoning**:
Now that `combinations` is imported and `co_authorship_pairs` is initialized, I will iterate through the `cleaned_authors` column, generate all unique pairs of authors for each entry, sort these pairs, and add them to the `co_authorship_pairs` list.



```python
for authors_list in df_filtered['cleaned_authors']:
    if len(authors_list) > 1:
        # Generate all unique pairs of authors for the current publication
        for pair in combinations(authors_list, 2):
            # Ensure the pair is sorted alphabetically to avoid duplicates like ('A', 'B') and ('B', 'A')
            co_authorship_pairs.append(tuple(sorted(pair)))

print(f"Total co-authorship pairs generated: {len(co_authorship_pairs)}")
print("First 10 co-authorship pairs:")
display(co_authorship_pairs[:10])
```

## Calculate Co-authorship Frequencies

### Subtask:
Count the number of times each unique pair of authors has collaborated.


**Reasoning**:
I need to import the `Counter` class to count the frequency of each co-authorship pair and then display the top 10 most frequent collaborations to verify the results.



```python
from collections import Counter

# Count the frequency of each co-authorship pair
co_authorship_frequency = Counter(co_authorship_pairs)

print("Top 10 most frequent co-authorship pairs:")
display(co_authorship_frequency.most_common(10))
```

## Build Co-authorship Network

### Subtask:
Construct a graph using a library like networkx, where authors are nodes and co-authorship forms edges.


**Reasoning**:
To construct the co-authorship network, I need to import the `networkx` library and initialize an empty undirected graph.



```python
import networkx as nx

G = nx.Graph()

print("Imported 'networkx' and initialized an empty graph G.")
```

**Reasoning**:
Now that the `networkx` graph `G` is initialized, I will iterate through the `co_authorship_frequency` to add edges representing collaborations to the graph, with the frequency as the edge weight.



```python
for pair, frequency in co_authorship_frequency.items():
    G.add_edge(pair[0], pair[1], weight=frequency)

print(f"Number of nodes in the graph: {G.number_of_nodes()}")
print(f"Number of edges in the graph: {G.number_of_edges()}")
```

## Visualize Co-authorship Network

### Subtask:
Visualize the co-authorship network to show collaboration patterns, with authors as nodes and collaborations as edges. Make sure to use appropriate layout algorithms and node/edge attributes to highlight key features.

**Reasoning**:
To visualize the network, I will use `matplotlib.pyplot` for plotting and `networkx.draw` for rendering the graph. It's often helpful to calculate centrality measures, such as degree centrality, to understand the importance of nodes (authors) in the network. The size of the nodes can then be scaled based on their centrality, making more influential authors stand out. I will also use a suitable layout algorithm for better readability, such as the Fruchterman-Reingold force-directed algorithm, and scale edge widths by collaboration frequency to visually represent stronger collaborations.

**Reasoning**:
Now, I will import `matplotlib.pyplot` and calculate degree centrality for each author to determine node sizes. I will then use `networkx.spring_layout` to arrange the nodes and `networkx.draw_networkx` to plot the graph, with node sizes reflecting centrality and edge widths reflecting collaboration frequency.



```python
import matplotlib.pyplot as plt

# Calculate degree centrality for node sizing (more collaborations = larger node)
degree_centrality = nx.degree_centrality(G)
node_size = [v * 2000 for v in degree_centrality.values()]

# Get edge weights for scaling edge width
edges = G.edges(data=True)
edge_weights = [d['weight'] for u, v, d in edges]
edge_width = [w * 0.5 for w in edge_weights] # Scale factor for better visibility

plt.figure(figsize=(18, 14))

# Use a spring layout for better visualization of clusters
pos = nx.spring_layout(G, k=0.15, iterations=50) # k regulates the distance between nodes

# Draw nodes
nx.draw_networkx_nodes(G, pos, node_size=node_size, node_color='skyblue', alpha=0.9)

# Draw edges with varying width based on collaboration frequency
nx.draw_networkx_edges(G, pos, width=edge_width, alpha=0.5, edge_color='gray')

# Draw labels for nodes (authors) - only for top 10% by degree centrality to avoid clutter
labels = {node: node for node in G.nodes() if degree_centrality[node] > sorted(degree_centrality.values(), reverse=True)[int(len(degree_centrality)*0.1)]}
nx.draw_networkx_labels(G, pos, labels, font_size=8, font_weight='bold')

plt.title('Co-authorship Network', size=20)
plt.axis('off')
plt.show()

print("Generated the co-authorship network visualization.")
```

```python
keyword_co_occurrence_pairs = []

for keywords_list in df_filtered['processed_keywords']:
    if len(keywords_list) > 1:
        # Generate all unique pairs of keywords for the current title
        for pair in combinations(keywords_list, 2):
            # Ensure the pair is sorted alphabetically to avoid duplicates like ('A', 'B') and ('B', 'A')
            keyword_co_occurrence_pairs.append(tuple(sorted(pair)))

print(f"Total keyword co-occurrence pairs generated: {len(keyword_co_occurrence_pairs)}")
print("First 10 keyword co-occurrence pairs:")
display(keyword_co_occurrence_pairs[:10])
```

## Calculate Keyword Co-occurrence Frequencies

### Subtask:
Count the number of times each unique pair of keywords co-occurs across all titles.


**Reasoning**:
To count the frequency of each keyword co-occurrence pair, I need to import the `Counter` class and apply it to the `keyword_co_occurrence_pairs` list. After calculating the frequencies, displaying the top 10 pairs will help verify the results.



```python
from collections import Counter

# Count the frequency of each keyword co-occurrence pair
keyword_co_occurrence_frequency = Counter(keyword_co_occurrence_pairs)

print("Top 10 most frequent keyword co-occurrence pairs:")
display(keyword_co_occurrence_frequency.most_common(10))
```

## Build Keyword Co-occurrence Network

### Subtask:
Construct a graph using `networkx` where keywords are nodes and co-occurrence forms edges, with the frequency as the edge weight.


**Reasoning**:
To construct the keyword co-occurrence network, I will initialize an empty `networkx` graph and then add edges for each keyword co-occurrence pair, using their frequency as the edge weight.



```python
import networkx as nx

G_keywords = nx.Graph()

for pair, frequency in keyword_co_occurrence_frequency.items():
    G_keywords.add_edge(pair[0], pair[1], weight=frequency)

print(f"Number of nodes in the keyword co-occurrence graph: {G_keywords.number_of_nodes()}")
print(f"Number of edges in the keyword co-occurrence graph: {G_keywords.number_of_edges()}")
```

## Visualize Keyword Co-occurrence Network

### Subtask:
Visualize the keyword co-occurrence network to show relationships between keywords, with keywords as nodes and co-occurrences as edges. Make sure to use appropriate layout algorithms and node/edge attributes to highlight key features.

**Reasoning**:
To visualize the network, I will use `matplotlib.pyplot` for plotting and `networkx.draw` for rendering the graph. It's often helpful to calculate centrality measures, such as degree centrality, to understand the importance of nodes (keywords) in the network. The size of the nodes can then be scaled based on their centrality, making more influential keywords stand out. I will also use a suitable layout algorithm for better readability, such as the Fruchterman-Reingold force-directed algorithm, and scale edge widths by co-occurrence frequency to visually represent stronger co-occurrences. Given the potential density, I will focus on labeling only the most central keywords to avoid clutter.

**Reasoning**:
Now, I will import `matplotlib.pyplot` and calculate degree centrality for each keyword to determine node sizes. I will then use `networkx.spring_layout` to arrange the nodes and `networkx.draw_networkx` to plot the graph, with node sizes reflecting centrality and edge widths reflecting co-occurrence frequency, labeling only the most central keywords.



```python
import matplotlib.pyplot as plt

# Calculate degree centrality for node sizing (more co-occurrences = larger node)
degree_centrality_keywords = nx.degree_centrality(G_keywords)
node_size_keywords = [v * 3000 for v in degree_centrality_keywords.values()] # Scale factor adjusted for keywords

# Get edge weights for scaling edge width
edges_keywords = G_keywords.edges(data=True)
edge_weights_keywords = [d['weight'] for u, v, d in edges_keywords]
edge_width_keywords = [w * 0.1 for w in edge_weights_keywords] # Scale factor for better visibility

plt.figure(figsize=(20, 16))

# Use a spring layout for better visualization of clusters
pos_keywords = nx.spring_layout(G_keywords, k=0.2, iterations=50) # k regulates the distance between nodes

# Draw nodes
nx.draw_networkx_nodes(G_keywords, pos_keywords, node_size=node_size_keywords, node_color='lightcoral', alpha=0.9)

# Draw edges with varying width based on co-occurrence frequency
nx.draw_networkx_edges(G_keywords, pos_keywords, width=edge_width_keywords, alpha=0.4, edge_color='darkgray')

# Draw labels for nodes (keywords) - only for top 10% by degree centrality to avoid clutter
labels_keywords = {node: node for node in G_keywords.nodes() if degree_centrality_keywords[node] > sorted(degree_centrality_keywords.values(), reverse=True)[int(len(degree_centrality_keywords)*0.1)]}
nx.draw_networkx_labels(G_keywords, pos_keywords, labels_keywords, font_size=9, font_weight='bold')

plt.title('Keyword Co-occurrence Network', size=24)
plt.axis('off')
plt.show()

print("Generated the keyword co-occurrence network visualization.")
```

## Validity of Analytical Approaches

### 1. Keyword Extraction Validity
The `extract_keywords` function used in this analysis effectively captures thematic content from titles by performing several crucial steps:
*   **Lowercasing**: Standardizes text to avoid issues with case sensitivity (e.g., "Terrorism" vs. "terrorism").
*   **Punctuation and Number Removal**: Cleans text by removing non-alphabetic characters, which are typically irrelevant for thematic analysis.
*   **Stop Word Filtering**: Removes common words (e.g., "the", "a", "is") that carry little semantic weight, thus focusing on more meaningful terms.
*   **Single-Character Word Filtering**: Eliminates very short words that are often noise or remnants of cleaning.

However, there are aspects that could be improved for deeper thematic analysis:
*   **Absence of Stemming/Lemmatization**: The current approach treats words like "analysis" and "analyzing" as distinct. Implementing stemming (reducing words to their root form, e.g., "connect", "connecting", "connected" -> "connect") or lemmatization (reducing words to their dictionary form, e.g., "better" -> "good") could group related terms, providing a more consolidated view of themes. This might reveal stronger co-occurrence patterns for core concepts that are currently diluted by different word forms.
*   **Contextual Nuance**: Keyword extraction from titles alone might miss thematic content present in abstracts or full text. Titles are concise and may not fully encompass all sub-themes.

Despite these points, for a quick overview of dominant terms and their co-occurrence, the current method is reasonable and provides a good first-pass understanding of explicit thematic content.

### 2. Centrality Measures Validity
**Degree Centrality** was used to assess the influence or connectivity of authors and keywords. This measure is appropriate for identifying immediate popularity or direct connections:
*   **Authors**: For authors, degree centrality represents the number of direct collaborators. A high degree central author is someone who has co-authored with many different individuals, suggesting a broad network. This is a valid measure for identifying prolific collaborators or 'network hubs'.
*   **Keywords**: For keywords, degree centrality indicates how many other keywords a particular term co-occurs with. A high degree central keyword is a broad concept that appears alongside many different topics, suggesting its foundational or cross-cutting nature within the dataset. This helps identify keywords that are central to multiple research discussions.

However, other centrality measures could offer different insights:
*   **Betweenness Centrality**: Measures the extent to which a node lies on the shortest paths between other nodes. For authors, this would identify 'brokers' who connect otherwise disconnected groups. For keywords, it would highlight concepts that bridge different thematic clusters.
*   **Closeness Centrality**: Measures how close a node is to all other nodes in the network. High closeness central authors/keywords can quickly interact with or spread information throughout the network.
*   **Eigenvector Centrality**: Assigns relative scores to all nodes in the network based on the concept that connections to high-scoring nodes contribute more to the score of the node in question. This can identify influential authors (those connected to other influential authors) or core keywords within a dominant theme.

Using only degree centrality provides a direct, easy-to-understand metric. For a more comprehensive understanding of network structure and influence, exploring these other measures would be beneficial.

## Enhance Keyword Extraction with Stemming

### Subtask:
Modify the `extract_keywords` function to include stemming using NLTK's Porter Stemmer. This will consolidate different forms of words (e.g., 'analysis', 'analyzing' to 'analys') to capture thematic content more effectively. This will require updating the `stop_words` set accordingly.


**Reasoning**:
To enhance keyword extraction, I will import NLTK's Porter Stemmer, initialize it, stem the existing stop words, and then redefine the `extract_keywords` function to incorporate stemming for all words before filtering out stop words and single-character words.



```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from collections import Counter
import re
import pandas as pd

# Download NLTK stopwords data if not already downloaded
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# 1. Initialize the PorterStemmer object
stemmer = PorterStemmer()

# Get the original stop words
stop_words = set(stopwords.words('english'))

# 2. Create a new stemmed_stop_words set
stemmed_stop_words = {stemmer.stem(word) for word in stop_words}

# Redefine the function to preprocess text and extract stemmed keywords
def extract_keywords(text):
    if pd.isna(text):  # Handle NaN values
        return []
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers
    words = text.split()

    # Apply stemmer to each word
    stemmed_words = [stemmer.stem(word) for word in words]

    # Filter out stemmed stopwords and single-character words
    keywords = [word for word in stemmed_words if word not in stemmed_stop_words and len(word) > 1]
    return keywords

print("Modified 'extract_keywords' function to include stemming and updated stemmed_stop_words set.")
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import re

# Load the original DataFrame (from cell bb87ef4f)
df = pd.read_csv('/content/citations (4).csv')

# Filter the DataFrame (from cell 0f4f13a0)
df_filtered = df[(df['Year'] >= 2012) & (df['Year'] <= 2022)]

# Download NLTK stopwords data if not already downloaded
try:
    stopwords.words('english')
except LookupError:
    nltk.download('stopwords')

# Initialize the PorterStemmer object (from cell 61df5384)
stemmer = PorterStemmer()

# Get the original stop words
stop_words = set(stopwords.words('english'))

# Create a new stemmed_stop_words set
stemmed_stop_words = {stemmer.stem(word) for word in stop_words}

# Redefine the function to preprocess text and extract stemmed keywords (from cell 61df5384)
def extract_keywords(text):
    if pd.isna(text):  # Handle NaN values
        return []
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text) # Remove punctuation and numbers
    words = text.split()

    # Apply stemmer to each word
    stemmed_words = [stemmer.stem(word) for word in words]

    # Filter out stemmed stopwords and single-character words
    keywords = [word for word in stemmed_words if word not in stemmed_stop_words and len(word) > 1]
    return keywords

# Apply the function to the 'Title' column to get processed_keywords (from cell 7615a50e)
df_filtered['processed_keywords'] = df_filtered['Title'].apply(extract_keywords)

# Join the lists of processed_keywords into single strings for each entry
joined_keywords = df_filtered['processed_keywords'].apply(lambda x: ' '.join(x))

# Initialize TfidfVectorizer
tfidf_vectorizer = TfidfVectorizer()

# Fit and transform the joined keyword strings
tfidf_matrix = tfidf_vectorizer.fit_transform(joined_keywords)

# Print the shape of the resulting TF-IDF matrix
print(f"Shape of TF-IDF matrix: {tfidf_matrix.shape}")
```

**Reasoning**:
The previous step successfully vectorized the keywords using TF-IDF. The next logical step in K-Means clustering is to determine the optimal number of clusters (K). I will achieve this by calculating the Silhouette score for a range of K values, which helps evaluate the quality of clusters.



```python
## Determine Optimal Number of Clusters (K)

### Subtask:
Determine the optimal number of clusters (K) for K-Means clustering using evaluation metrics like the Silhouette score. This step is crucial for applying K-Means effectively.

**Reasoning**:
To determine the optimal number of clusters (K) for K-Means, I will use the Silhouette score, which measures how similar an object is to its own cluster (cohesion) compared to other clusters (separation). I'll iterate through a range of potential cluster numbers (e.g., 2 to 10), apply K-Means for each, and calculate the Silhouette score. The K value that yields the highest Silhouette score will be considered optimal. This approach helps in finding a well-defined clustering structure.
```

**Reasoning**:
The previous code block failed because it contained markdown text within a Python code cell, leading to a `SyntaxError`. I need to provide the correct Python code to determine the optimal number of clusters (K) using the Silhouette score, which involves iterating through a range of K values, applying K-Means, and calculating the Silhouette score for each.



```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Define a range of K values to test
k_range = range(2, 11) # Testing from 2 to 10 clusters
silhouette_scores = []

# Iterate through the K values, apply K-Means, and calculate Silhouette score
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) # n_init for robust centroid initialization
    kmeans.fit(tfidf_matrix)
    cluster_labels = kmeans.labels_

    # Ensure there's more than one cluster and more than one sample for silhouette_score
    if len(set(cluster_labels)) > 1 and len(cluster_labels) > 1:
        score = silhouette_score(tfidf_matrix, cluster_labels)
        silhouette_scores.append(score)
    else:
        silhouette_scores.append(-1) # Indicate an issue if clustering is degenerate

# Plot the Silhouette scores to find the optimal K
plt.figure(figsize=(10, 6))
plt.plot(k_range, silhouette_scores, marker='o')
plt.title('Silhouette Score for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(list(k_range))
plt.grid(True)
plt.show()

# Find the optimal K (excluding -1 values if any)
optimal_k = k_range[silhouette_scores.index(max(silhouette_scores))] if max(silhouette_scores) != -1 else None

print(f"Silhouette scores for K from 2 to 10: {silhouette_scores}")
print(f"Optimal number of clusters (K) based on Silhouette score: {optimal_k}")
```

**Reasoning**:
Now that the optimal number of clusters (K) has been determined, I will apply the K-Means algorithm to the `tfidf_matrix` using this optimal K. This will cluster the documents (titles) based on their keyword content.



```python
from sklearn.cluster import KMeans

# Apply K-Means with the optimal K
kmeans_model = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
cluster_labels = kmeans_model.fit_predict(tfidf_matrix)

# Add cluster labels to the df_filtered DataFrame
df_filtered['cluster_label'] = cluster_labels

print(f"K-Means clustering applied with optimal K = {optimal_k}.")
print("First 10 cluster labels:")
display(cluster_labels[:10])
print("Counts per cluster:")
display(df_filtered['cluster_label'].value_counts().sort_index())
```
